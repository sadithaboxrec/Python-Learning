### Write a Python program that accepts a full name and prints the initials in uppercase.

```python
full_name = input("Enter your full name: ")

# Split the name into separate words
parts = full_name.split()

initials = ""

# Go through each word in the name
for word in parts:
    first_letter = word[0].upper()
    initials += first_letter


print("Initials:", initials)
```

## Write a Python script that takes a user's full name and reverses each word separately.

```python
full_name = input("Enter your full name: ")

name_parts = full_name.split()
reversed_name_parts = []

for word in name_parts:
    reversed_word = word[::-1]

    # Add the reversed word to the list
    reversed_name_parts.append(reversed_word)


reversed_name = " ".join(reversed_name_parts)

print("Reversed name:", reversed_name)
```
