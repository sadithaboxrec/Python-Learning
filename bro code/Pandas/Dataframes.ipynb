### DataFrame = A tabular data structure with rows and columns

```python
import pandas as pd
```

```python
data = {"Name":["Spongebob","Patrick","Squidward"],
        "Age":[30,35,50]
        }

df = pd.DataFrame(data)  # constructor
print(df)
df
```

```python
df=pd.DataFrame(data,index=['a','b','c'])
df
```

```python
print(df.loc['a'])
df.loc['b']
```

```python
df.iloc[0]
```

#### Add a new Column

```python
df["Job"] = ["Cook","N/A","Cashier"]
df
```

#### Add a new Row

```python
new_row = pd.DataFrame([{"Name":"Sandy","Age":28,"Job":"Scientist"}],index=['d'])

# concatenate it with existing dataframe
df=pd.concat([df,new_row])
df
```

#### Add new Rows

```python
new_rows = pd.DataFrame([{"Name":"Crabs","Age":60,"Job":"Manager"},
                        {"Name":"Snoopy","Age":15,"Job":"Writer"}], index=['e','f']
                        )

df=pd.concat([df,new_rows])
df
```
