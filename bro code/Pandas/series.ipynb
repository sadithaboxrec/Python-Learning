### Series

##### 1 dimensional labeled array that can hold any data type

```python
data=[100,102,104]

series=pd.Series(data)   # Series is a constructor
print(series)
series
```

```python
data1=[100.1,102.2,104.3]

series1=pd.Series(data1)   # Series is a constructor
print(series1)
series1
```

```python
data2=["A","B","C"]

series2=pd.Series(data2)   # Series is a constructor
print(series2)
series2
```

```python
data3=[100,102,104]

series3=pd.Series(data3 ,index=["a","b","c"])   # Series is a constructor
print(series3)
series3
```

#### If any time need to access value directly with in a Series we can use loc function

```python
# series3.loc["a"]
print(series3.loc["a"])
```

##### to update value

```python
series3.loc["c"]  = 200
print(series3.loc["c"])
```

#### locate by integer

```python
print(series3.iloc[0])
print(series3.iloc[1])
print(series3.iloc[2])
```

## Filter by values

```python
data4=[100,102,104,200,202,206]

series4=pd.Series(data4 ,index=["a","b","c","d","e","f"])   # Series is a constructor
print(series4)
series4
```

```python
print(series4[series4>=200])
series4[series4>=200]
```

```python
calories = {"Day 1":1750,
            "Day 2":2100,
            "Day 3":1700 }

series5=pd.Series(calories)
series5
```

```python
print(series5.loc["Day 1"])
print(series5.loc["Day 2"])
```

```python
series5.loc["Day 2"] += 230
print(series5.loc["Day 2"])
```

```python
print(series5)
```

```python
print(series5[series5>=2000])
```

```python
print(series5[series5<2000])
```
