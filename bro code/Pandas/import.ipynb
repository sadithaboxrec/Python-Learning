## Select Multiple columns

```python
print(df[['Name','Height','Weight']].to_string())
```

```python
print(df[['Name','Height','Weight']])
```

### Selection by Rows

#### Each row thar row has a lable. default is given a number

```python
print(df.loc[0])
```

#### we can set columns as indexes too

```python
df=pd.read_csv("pokemon.csv",index_col="Name")
print(df.loc["Pikachu"])
```

```python
print(df.loc["Charizard",["Height","Weight"]])      # passing a second arguement
```

### range of rows

```python
print(df.loc["Charizard":"Blastoise",["Height","Weight"]])      # passing a second arguement
```

```python
print(df.iloc[0:13])      # passing a second arguement
```

```python
print(df.iloc[0:13:2])
```

```python
print(df.iloc[0:13:2,0:3])
```
