### The process of fixing removing incomplete incorrect data

```python
import pandas as pd
df=pd.read_csv("pokemon.csv")
print(df.head())
```

```python
df=df.drop(columns=['Legendary',"No"])      # Returns a new DataFrame , so assigned it to a value
print(df.head())
```

### Handles Missing Data

```python
df= df.dropna(subset=['Type2'])       # Drop Not Available
print(df.to_string())
```

```python
df=pd.read_csv("pokemon.csv")
print(df)
```

```python
df=df.fillna({"Type2":"None"})
print(df.to_string())
```

## Fix inconsistant values

```python
df['Type1']=df['Type1'].replace({'Grass':'GRASS'})
print(df.to_string())
```

```python
df['Type1']=df['Type1'].replace({'Water':'WATER',"Fire":"FIRE"})
print(df.to_string())
```

### Standardize Texts

```python
df['Name']=df['Name'].str.lower()
print(df)
```

## Fix data types


```python
df['Legendary']=df['Legendary'].astype(bool)
print(df)
```
