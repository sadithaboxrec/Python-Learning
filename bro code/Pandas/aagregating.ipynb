### Aggregate Functions    - Reduce a set of values into a single summary value used to summarize and analyze data often used with                               the groupby()  function

```python
import pandas as pd
df=pd.read_csv('pokemon.csv')
print(df.head())
```

### Applying to the whole Data Frame

```python
print(df.mean(numeric_only=True))
```

```python
print(df.sum(numeric_only=True))
```

```python
print(df.min(numeric_only=True))
```

```python
print(df.max(numeric_only=True))
```

```python
print(df.count())     # when there is no type 2 or NaN it will be ignored
```

### Single Column

```python
print(df["Height"].mean())
```

```python
print(df["Height"].sum())
```

```python
print(df["Height"].min())
```

```python
print(df["Height"].max())
```

```python
print(df["Height"].count())
```

```python
print(df["Type2"].count())
```

### Group By

```python
group = df.groupby('Type1')
print(group)
```

```python
group = df.groupby('Type1')
print(group["Height"].mean())
```

```python
print(group["Height"].sum())
```

```python
print(group["Height"].count())
```

```python
print(group["Type2"].max())
```
