## Logistic Regression

### Work Flow of Logistic Regression

#### 1] Set Learning Rate and Number of iteration, Initiate Random Weights and Bias values
#### 2] Build Logistic Regression Function (Sigmoid Function)
#### 3] Update the parameters using Gradient Descent. Get the best model with best weight and bias value as it has minimum cost function
#### 4] Build the predict function to determine the class of the data point

```python
class Logistic_Regression():

    # inititating the hyper parameters
    def __init__(self, learning_rate, no_of_iterations):
        self.learning_rate = learning_rate
        self.no_of_iterations = no_of_iterations

    # fit dataset we have for logistic regression to train
    def fit(self, X, Y):

         # number of data points in the dataset(number of rows) --> m
         # number of input features in the dataset (number of columns) --> n
        self.m, self.n = X.shape   # (769 * 8)

         # initiating weight and bias value
        self.W = np.zeros(self.n)
        self.b = 0

        self.X = X
        self.Y = Y

        # implementing gradient descent for optimization

        for i in range(self.no_of_iterations):
            self.update_weight()

    def update_weight(self):

        # Y_hat formula (sigmoid function)
        Y_hat = 1 / (1 + np.exp( - (self.X.dot(self.w) + self.b ) ))   # dot represents product of matrices

        # derivaties // gradients

        dw = (1/self.m)*np.dot(self.X.T, (Y_hat - self.Y))

        db = (1/self.m)*np.sum(Y_hat - self.Y)

        # updating the weights & bias using gradient descent

        self.w = self.w - self.learning_rate * dw
        self.b = self.b - self.learning_rate * db

    def predict(self, X):
        Y_pred = 1 / (1 + np.exp( - (X.dot(self.w) + self.b ) ))
        Y_pred = np.where( Y_pred > 0.5, 1, 0)
        return Y_pred
```
