```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

# Label Encoding of Breast Cancer Dataset

```python
# loading the data from csv file to pandas dataframes

cancer_data = pd.read_csv('../data/encode/data.csv')
```

```python
cancer_data.head()
```

```python
# M is - running stage
# B is - critical stage
```

```python
# finding the different  labels's count

cancer_data['diagnosis'].value_counts()
```

```python
# load the label encoder function

label_encoder = LabelEncoder()
```

```python
labels = label_encoder.fit_transform(cancer_data.diagnosis)
```

```python
# appending the labels to the data frame

cancer_data['target'] = labels
```

```python
cancer_data.head()
```

##  0 -->  Benign

##  1 --> Malignant


### //  when use label encode function label present in particular column will be arranged alphabatically

```python
cancer_data['target'].value_counts()
```
