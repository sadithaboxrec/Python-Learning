#### 1. Mean

#### analyze distribution of data before the select method to handle data

```python
fig, ax = plt.subplots(figsize=(8, 8))
sns.displot(dataset.salary)
```

### with a skew distribution like above with one side , we ain't use mean value, use median or mode

```python
dataset['salary'].fillna(dataset['salary'].median(), inplace=True)
```

```python
dataset['salary'].isnull().sum()
```
